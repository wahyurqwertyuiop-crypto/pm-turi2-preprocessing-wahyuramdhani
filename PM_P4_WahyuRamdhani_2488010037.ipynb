```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA', 'S1', 'SMA', 'S2', 'SMA', 'S2', 'S1', 'S1', 'SMA', 'S2'],  # ordinal
    'kota'      : ['Bandung', 'Jakarta', 'Bandung', 'Surabaya', 'Jakarta',
                   'Surabaya', 'Jakarta', 'Bandung', 'Bandung', 'Jakarta'],    # nominal
    'membeli'   : ['Tidak', 'Ya', 'Tidak', 'Ya', 'Tidak', 'Ya', 'Ya', 'Tidak', 'Tidak', 'Ya']  # target
}
df = pd.DataFrame(data)

df['status_pekerjaan'] = ['Tetap', 'Kontrak', 'Tetap', 'Tetap', 'Kontrak',
                          'Tetap', 'Kontrak', 'Tetap', 'Kontrak', 'Tetap']

print("---------DataFrame Awal----------")
print(df)
print("\n" + "="*60 + "\n")

df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())

X = df.drop(columns=['membeli'])
y = df['membeli']

X['pendidikan'] = X['pendidikan'].map({'SMA': 0, 'S1': 1, 'S2': 2})

X = pd.get_dummies(X, columns=['kota', 'status_pekerjaan'], dtype=int)

y = y.map({'Tidak': 0, 'Ya': 1})

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

num = ['usia', 'pendapatan', 'pendidikan']
sc = MinMaxScaler()

X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])

print("--- Deskripsi Statistik Data Latih (X_train) setelah MinMaxScaler ---")
print(X_train.describe())

print("\n------ Contoh Data Latih (X_train) Terakhir -------")
print(X_train.head())

print("\nD.4")
print("Metode fit_transform digunakan pada data latih (X_train) untuk menghitung parameter statistik (seperti rata-rata, standar deviasi, nilai min/max) sekaligus melakukan transformasi data tersebut. Sedankan Data uji (X_test) hanya boleh di-transform menggunakan parameter statistik yang sama dari data latih. Jika kita melakukan fit ulang pada data uji, model akan mempelajari parameter dari data uji itu sendiri, yang dapat menyebabkan kebocoran informasi (data leakage) dan membuat performa evaluasi model menjadi tidak valid atau terlalu optimis.")

print("\nE.1")
print("Urutan ini sangat penting untuk mencegah terjadinya data leakage (kebocoran data) dari data uji ke dalam data latih. Jika kita melakukan scaling atau imputasi pada seluruh dataset sebelum splitting, informasi statistik dari data uji (seperti nilai maksimum atau rata-rata) akan ikut tercampur saat proses fit, sehingga model seolah-olah melihat data masa depan yang seharusnya tidak boleh diketahui selama proses pelatihan.")

print("\nE.2")
print("Label Encoding digunakan untuk variabel kategorikal yang bersifat ordinal (memiliki tingkatan atau urutan yang bermakna, contohnya: Rendah = 0, Sedang = 1, Tinggi = 2).")
print("One-Hot Encoding digunakan untuk variabel kategorikal yang bersifat nominal (tidak memiliki tingkatan atau urutan, contohnya nama kota: Bandung, Jakarta, Surabaya) agar model tidak keliru mengasumsikan adanya hubungan matematis atau urutan antar kategori tersebut")

print("\nE.3")
print("Normalisasi (MinMaxScaler) mengubah skala data sehingga berada pada rentang nilai tertentu, umumnya antara 0 sampai 1. Cocok digunakan saat algoritma tidak berasumsi bentuk distribusi data tertentu (seperti Neural Networks atau K-Nearest Neighbors).")
print("Standardisasi (StandardScaler) mengubah skala data agar memiliki rata-rata (mean) sama dengan 0 dan simpangan baku (standard deviation) sama dengan 1 (distribusi normal standar). Cocok digunakan untuk algoritma yang berasumsi data berdistribusi normal (seperti Regresi Linear atau Support Vector Machines).")
```
